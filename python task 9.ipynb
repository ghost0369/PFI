```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Open"].mean()
```

```python
df["Open"] = df ["Open"].fillna (df ["Open"].mean())
```

```python
df["High"]=df["High"].fillna(df["High"].mean())
```

```python
df["Close"]=df["Close"].fillna (df["Close"].mean())
```

```python
df["Volume"]=df["Volume"].fillna (df ["Volume"].mean())
```

```python
df.isnull().sum()
```

```python
df["Data"]=pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
plt.figure(figsize=(19, 7))
sns.barplot(x="Date", y="Volume", data=df, width=0.6,color="Yellow")
plt.title("Trending Volume over Time")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.grid(True)
plt.xticks (rotation=45)
plt.show()
```

```python
plt.figure(figsize=(19, 7))
sns.lineplot(x="Date", y="Close", data=df,color="Blue")
plt.title("Stock Closing")
plt.xlabel("Date")
plt.ylabel("Close")
plt.grid(True)
plt.xticks (rotation=45)
plt.show()
```

```python
correlation=df[["Open", "High", "Low", "Close", "Volume"]].corr()
```

```python
correlation
```

```python
sns.heatmap(correlation,
annot=True, cmap="Greens",
cbar=False,
square=True,
linewidth=2)
plt.show()
```

```python
df["Budgrt"]=df["Close"]
df["Actual"]=df["Open"]
```

```python
data = {
    "department": ["Sales", "Engineering", "Marketing", "HR","Intern","Pa"],
    "budget": [60000, 120000, 70000, 67000, 20000,25000],
    "actual": [49000, 100000, 58000, 30000,15000,20000],
}

df = pd.DataFrame(data)
print(df)

df_long=df.melt(
    id_vars="department",
    value_vars=["budget","actual"],
    var_name="Type",
    value_name="Amount"
)



sns.barplot(
    data=df_long,
    x="department",
    y="Amount",
    hue="Type",
    color="Blue",
)
plt.title("Budget vs Actual")
plt.xlabel("Department")
plt.ylabel("Amount")
plt.show()
```
