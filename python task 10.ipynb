```python
df["Data"]=pd.to_datetime(df["Date"])
```

```python
df=df.sort_values("Date")
```

```python
df ["MA_20"]=df["Close"].rolling (20).mean()
df["MA_50"]=df["Close"].rolling(50).mean()
```

```python
df.head()
```

```python
plt.figure(figsize=(17,11))
plt.plot(df["Date"], df["Close"], label="Closing price", marker="o", color="Blue")
plt.plot(df["Date"], df["MA_20"], label="Moving avrage of 20 days", marker="o", color="Black")
plt.plot(df["Date"], df["MA_50"], label="Moving avrage of 50 days", marker="o", color="Yellow")
plt.title("Stock price trends with moving avarge")
plt.grid(True)
plt.ylabel("Price")
plt.xlabel("Date")
plt.legend()
plt.xticks (rotation=35)
plt.show()
```
