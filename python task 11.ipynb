```python
total_sales = df["Sales"].sum()
total_expenses = df["Expenses"].sum()
total_Profit = df["Profit"].sum()
total_assets = df["Assets"].sum()
total_liabilities = df["Liabilities"].sum()

print("Total_sales:", total_sales)
print("Total_expenses:", total_expenses)
print("Total_Profit:", total_Profit)
print("Total Assets:", total_assets)
print("Total Liabilities:", total_liabilities)
```

```python
Average_sales = df["Sales"].mean()
highest_Sales = df["Sales"].max()
average_stock_price = df["Stock_Price"].mean()

print("Average_sales:", Average_sales)
print("highest Sales:", highest_Sales)
print("Average Stock Price:", average_stock_price)
```

```python
monthly_sales=df.groupby("Month") [["Sales", "Expenses", "Profit"]].mean()
print("Monthly Financial Summary:-")
print(monthly_sales)
```

```python
plt.figure(figsize=(16, 7))
plt.plot(
    monthly_sales.index,
    monthly_sales["Profit"],
    marker="o",
    alpha=0.7,
    label="Profit",
)
plt.plot(
    monthly_sales.index,
    monthly_sales["Sales"],
    marker="o",
    alpha=0.7,
    label="Sales",
)
plt.xlabel("Month")
plt.ylabel("Amount")
plt.title("Monthly Sales and Profit")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python

yearly_summary = df.groupby("Year")[["Sales", "Expenses", "Profit"]].sum()

plt.figure(figsize=(14, 7))
plt.plot(
    yearly_summary.index,
    yearly_summary["Sales"],
    marker="o",
    label="Sales",
)
plt.xlabel("Year")
plt.ylabel("Sales")
plt.title("Yearly Sales")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(16,7))
sns.barplot(x="Month", y="Sales", data=df,width=0.6)
plt.title("category vs profit")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("monthly sales")
plt.xticks (rotation=45)
plt.show()
```

```python
profit_margin= ((df["Profit"]) / df ["Sales"])*100
print("Profit Margin:", profit_margin)
```

```python
plt.figure(figsize=(16, 7))
sns.scatterplot(x="Sales", y="Profit", data=df, hue="Sales")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```
