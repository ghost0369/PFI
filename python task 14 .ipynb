```python
df["Length (Duration)"] = pd.to_numeric(
    df["Length (Duration)"],
    errors="coerce"
)

print(df["Length (Duration)"].dtype)
```

```python
average_popularity = df["Popularity"].mean()

print("Average Popularity:",(average_popularity))
```

```python
maximum_popularity = df["Popularity"].max()

print("Maximum Popularity:", maximum_popularity)
```

```python
average_bpm = df["Beats Per Minute (BPM)"].mean()

print("Average BPM:", (average_bpm))
```

```python
average_valence = df["Valence"].mean()

print("Average Valence:",(average_valence))
```

```python
average_duration = df["Length (Duration)"].mean()

print("Average Duration:",(average_duration), "seconds")
```

```python
genre_summary = df.groupby("Top Genre").agg(
    Songs=("Title", "count"),
    Avg_Popularity=("Popularity", "mean"),
    Avg_Energy=("Energy", "mean")
).sort_values("Songs", ascending=False)

genre_summary.head(10)
```

```python
artist_summary = df.groupby("Artist").agg(
    Songs=("Title", "count"),
    Avg_Popularity=("Popularity", "mean"),
    Avg_Danceability=("Danceability", "mean")
).sort_values("Songs", ascending=False)

artist_summary.head(10)
```

```python
year_summary = df.groupby("Year").agg(
    Songs=("Title", "count"),
    Avg_Popularity=("Popularity", "mean"),
    Avg_Energy=("Energy", "mean")
).sort_index()

year_summary.head(10)
```

```python
top_songs = df.nlargest(10, "Popularity")

plt.figure(figsize=(12, 6))

plt.barh(
    top_songs["Title"],
    top_songs["Popularity"]
)

plt.xlabel("Popularity")
plt.ylabel("Song")
plt.title("Top 10 Most Popular Songs")

plt.gca().invert_yaxis()

plt.show()
```

```python
top_genres = df["Top Genre"].value_counts().head(10)

plt.figure(figsize=(12, 6))

top_genres.plot(kind="bar")

plt.xlabel("Genre")
plt.ylabel("Number of Songs")
plt.title("Top 10 Genres by Number of Songs")

plt.xticks(rotation=45)

plt.show()
```

```python
songs_by_year = df.groupby("Year")["Title"].count()

plt.figure(figsize=(12, 6))

plt.plot(
    songs_by_year.index,
    songs_by_year.values
)

plt.xlabel("Year")
plt.ylabel("Number of Songs")
plt.title("Number of Songs by Year")

plt.grid(True)

plt.show()
```

```python
popularity_by_year = df.groupby("Year")["Popularity"].mean()

plt.figure(figsize=(12, 6))

plt.plot(
    popularity_by_year.index,
    popularity_by_year.values
)

plt.xlabel("Year")
plt.ylabel("Average Popularity")
plt.title("Average Song Popularity by Year")

plt.grid(True)

plt.show()
```

```python
top_genre_energy = genre_summary.head(10)

plt.figure(figsize=(12, 6))

plt.bar(
    top_genre_energy.index,
    top_genre_energy["Avg_Energy"]
)

plt.xlabel("Genre")
plt.ylabel("Average Energy")
plt.title("Average Energy of Top 10 Genres")

plt.xticks(rotation=45)

plt.show()
```

```python
top_artist_dance = artist_summary.head(10)

plt.figure(figsize=(12, 6))

plt.bar(
    top_artist_dance.index,
    top_artist_dance["Avg_Danceability"]
)

plt.xlabel("Artist")
plt.ylabel("Average Danceability")
plt.title("Average Danceability of Top 10 Artists")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Danceability"],
    df["Popularity"],
    alpha=0.6
)

plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.title("Popularity vs Danceability")

plt.grid(True)

plt.show()
```
